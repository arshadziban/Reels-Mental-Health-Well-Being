```python
import os
import pandas as pd
from dotenv import load_dotenv
from gsheet_loader import load_sheet
load_dotenv()
```

```python
# Load a Google Sheet
url = os.environ["GOOGLE_SHEET_URL"]
data = load_sheet(url)
df = pd.read_csv(data)
df.head()
```

```python
# Rename columns to short, code-based names (e.g. A1_age, A2_gender, ...)
import re

# Human-readable label for each question code (based on the English question text)
code_labels = {
    "A1": "age",
    "A2": "gender",
    "A3": "religion",
    "A4": "marital_status",
    "A5": "residence",
    "A6": "occupation",
    "B1": "family_type",
    "B2": "family_time_hours",
    "B3": "family_relationship",
    "B4": "in_person_feeling",
    "C1": "height",
    "C2": "weight",
    "C3": "sitting_duration",
    "C4": "physical_problems",
    "D1": "years_watching",
    "D2": "daily_watch_time",
    "D3": "peak_watch_time",
    "D4": "multitasking_while_watching",
    "E1": "sleep_hours",
    "E2": "sleep_trouble",
    "E3": "exercise_involvement",
    "E4": "substance_use",
    "F1": "attention_decreased",
    "F2": "memory_reduced",
    "F3": "learning_interest_decreased",
    "F4": "performance_affected",
    "G1": "mood_swings",
    "G2": "fomo_anxiety",
    "G3": "social_comparison",
    "G4": "social_isolation",
    "G5": "mental_stress",
    "G6": "panic_attacks",
    "G7": "behavior_type",
    "G8": "addictive_watching",
    "H1": "mental_state_change",
    "H2": "content_type_effect",
    "H3": "distraction_after_viral",
    "H4": "brain_pressure",
    "H5": "sleep_quality_change",
    "H6": "emotional_reels_focus_loss",
    "H7": "curiosity_driven_rewatch",
}

code_pattern = re.compile(r"^\s*([A-H]\d{1,2})\s*[.)]")


def clean_column_name(col: str) -> str:
    if col.strip().lower() == "timestamp":
        return "timestamp"
    match = code_pattern.match(col)
    if match:
        code = match.group(1)
        label = code_labels.get(code)
        if label:
            return f"{code}_{label}"
    return col


df = df.rename(columns=clean_column_name)
df = df.drop(columns=["timestamp"])
df.columns.tolist()
```

```python
# Remove Bangla text, keeping only the English part of each value
# Values look like "English text (বাংলা text)" -> keep "English text"
bangla_pattern = re.compile(r"\s*\([^()]*[ঀ-৿][^()]*\)")


def strip_bangla(value):
    if isinstance(value, str):
        cleaned = bangla_pattern.sub("", value)
        cleaned = re.sub(r"[ঀ-৿]+", "", cleaned)
        return cleaned.strip()
    return value


df = df.apply(lambda col: col.map(strip_bangla))
df.head()
```

```python
# Show all unique values (with counts) for every column as an HTML report
from html import escape
from IPython.display import HTML, display


def describe_value(value):
    """Return (label, css_class) so missing / blank / junk values stand out."""
    if pd.isna(value):
        return "NaN (missing)", "missing"
    text = str(value)
    if text.strip() == "":
        return f"'{text}' (blank)", "missing"
    if not re.search(r"[A-Za-z0-9]", text):
        return f"'{text}' (no letters/digits)", "junk"
    return text, ""


sections = []
for col in df.columns:
    counts = df[col].value_counts(dropna=False)
    n_missing = int(df[col].isna().sum() + (df[col].astype(str).str.strip() == "").sum())
    rows = "".join(
        f"<tr class='{cls}'><td>{escape(label)}</td><td class='num'>{count}</td></tr>"
        for label, cls, count in (
            (*describe_value(value), count) for value, count in counts.items()
        )
    )
    sections.append(
        f"<details {'open' if len(counts) <= 10 else ''}>"
        f"<summary><b>{escape(col)}</b> &mdash; {len(counts)} unique, "
        f"{n_missing} missing/blank</summary>"
        f"<table><tr><th>Value</th><th>Count</th></tr>{rows}</table></details>"
    )

style = """
<style>
.uv details { margin: 6px 0; border: 1px solid #8884; border-radius: 6px; padding: 6px 10px; }
.uv summary { cursor: pointer; }
.uv table { border-collapse: collapse; margin-top: 6px; min-width: 360px; }
.uv th, .uv td { border: 1px solid #8884; padding: 3px 8px; text-align: left; white-space: pre-wrap; }
.uv td.num { text-align: right; width: 60px; }
.uv tr.missing td { background: #e5484d33; }
.uv tr.junk td { background: #f5a52433; }
</style>
"""
display(HTML(
    style
    + f"<div class='uv'><p>{len(df)} rows &times; {len(df.columns)} columns. "
    "Red = missing/blank, orange = no letters/digits. "
    "Columns with more than 10 values are collapsed &mdash; click to expand.</p>"
    + "".join(sections)
    + "</div>"
))
```

## Value Handeling

```python
# Cleaning helpers shared by every column below
# Rules: missing / unclear -> column mode | number + unit -> the number | sentences are NOT parsed (mode instead)
import numpy as np

df_before = df.copy()   # snapshot, used for the final comparison
clean_log = {}          # per-column statistics, filled by the cleaning functions

PLACEHOLDERS = {"", "n/a", "na", "n.a.", "null", "nan", "unknown", "not applicable", "not available", "nil"}


def is_missing_like(value):
    """NaN, blank string or a placeholder such as 'N/A' / 'unknown'."""
    return bool(pd.isna(value)) or str(value).strip().lower() in PLACEHOLDERS


def detect_type(series):
    values = series[~series.map(is_missing_like)].astype(str).str.strip()
    if set(values.str.lower().unique()) <= {"yes", "no", "true", "false"}:
        return "boolean"
    if values.nunique() <= 15:
        return "categorical"
    return "numerical" if values.str.contains(r"\d").mean() > 0.5 else "text"


def count_modified(before, after):
    if pd.api.types.is_numeric_dtype(after):
        same = pd.to_numeric(before.astype(str).str.strip(), errors="coerce").eq(after)
        return int((~same).sum())
    return int((before.astype("string").str.strip().fillna("<NA>") != after.astype("string")).sum())


def log_cleaning(col, ctype, before, after, missing, invalid, converted, mode):
    clean_log[col] = {
        "type": ctype,
        "modified": count_modified(before, after),
        "missing_replaced": int(missing),
        "invalid_replaced": int(invalid),
        "mixed_converted": int(converted),
        "mode_used": mode,
    }


def clean_categorical(col, valid=None):
    """Trim/standardise text, send missing + unclear values to the mode. Valid text is never changed."""
    raw = df[col]
    ctype = detect_type(raw)
    text = raw.astype("string").str.strip().str.replace(r"\s+", " ", regex=True)

    missing = raw.map(is_missing_like).astype(bool)
    # unclear = no letter/digit at all (e.g. "।", ",", "-"), or not one of the allowed values
    invalid = ~missing & ~text.str.contains(r"[A-Za-z0-9]", na=False).astype(bool)
    if valid is not None:
        invalid |= ~missing & ~text.str.lower().isin([v.lower() for v in valid]).astype(bool)

    text = text.mask(missing | invalid)
    key = text.str.lower()
    # one spelling per value = its most frequent spelling (Male / male / MALE -> Male)
    canonical = text.groupby(key).agg(lambda x: x.value_counts().idxmax())
    text = key.map(canonical)

    mode = text.mode().iloc[0]
    cleaned = text.fillna(mode).astype(str)
    df[col] = cleaned
    log_cleaning(col, ctype, raw, cleaned, missing.sum(), invalid.sum(), 0, mode)

    print(f"{col} | detected type: {ctype}")
    print(f"missing replaced: {missing.sum()} | invalid replaced: {invalid.sum()} | "
          f"modified: {clean_log[col]['modified']} | mode: {mode}")
    if invalid.any():
        print("invalid values:", raw[invalid].unique().tolist())
    summary = cleaned.value_counts().rename("count").to_frame()
    summary["percent"] = (summary["count"] / summary["count"].sum() * 100).round(1)
    display(summary)


# ---- numeric parsing -------------------------------------------------------
NUMBER = r"\d+(?:\.\d+)?"
QUALIFIERS = r"(?:(?:about|around|approximately|approx|almost|nearly|over|average|avg)\s+)?"
ZERO_WORDS = {"zero", "never", "none", "nil"}

# alias -> factor that converts it into the column's target unit
HOURS = {**dict.fromkeys(["h", "hr", "hrs", "hour", "hours", "houre", "hour's", "hours's", "gonta", "ghonta"], 1.0),
         **dict.fromkeys(["m", "min", "mins", "minute", "minutes"], 1 / 60)}
YEARS = {**dict.fromkeys(["y", "yr", "yrs", "year", "years", "yeasr"], 1.0),
         **dict.fromkeys(["month", "months"], 1 / 12),
         **dict.fromkeys(["week", "weeks"], 1 / 52)}
KILOS = dict.fromkeys(["kg", "kgs", "kilo", "kilos", "kilogram", "kilograms"], 1.0)


def make_quantity_parser(units, min_value, max_value):
    """Strict parser: the WHOLE value must be '<number> [unit]', a range ('5-6', '5/6') or 'N unit M unit'.
    Anything else (sentences, opinions, numbers hidden in text) is invalid -> None."""
    alias = "|".join(re.escape(u) for u in sorted(units, key=len, reverse=True))
    pattern = re.compile(
        rf"{QUALIFIERS}({NUMBER})(?:\s*(-|–|/|\\|to)\s*({NUMBER}))?\s*({alias})?(?:\s+({NUMBER})\s*({alias}))?"
    )

    def parse(value):
        t = str(value).strip().lower().rstrip(". ")
        if t in ZERO_WORDS:
            return 0.0, "converted"
        m = pattern.fullmatch(t)
        if not m:
            return None, "invalid"
        a, sep, b, unit, c, unit2 = m.groups()
        a, f = float(a), units[unit] if unit else 1.0
        if b:
            if sep == "/" and (a, float(b)) == (1.0, 2.0):      # "1/2 hour" - half an hour or 1-2 hours?
                return None, "invalid"
            amount, kind = (a + float(b)) / 2 * f, "range"
        elif c:
            amount, kind = a * f + float(c) * units[unit2], "converted"
        else:
            amount, kind = a * f, ("plain" if re.fullmatch(NUMBER, t) else "converted")
        return (amount, kind) if min_value <= amount <= max_value else (None, "invalid")

    return parse


FT = r"(?:'|\"|ft|feet|foot|fit)"
INCH = r"(?:\"|inches|inche|inch|inc|in)"


def parse_height_cm(value):
    """Height in cm. Understands cm, feet/inches (5'6", 5 feet 6 inch, 5.6) and inches-only (69 inch)."""
    t = str(value).strip().lower()
    for q in "’‘`´′":
        t = t.replace(q, "'")
    for q in "”“″":
        t = t.replace(q, '"')
    t = t.replace("''", '"').rstrip(". ")

    kind = "plain" if re.fullmatch(NUMBER, t) else "converted"
    cm = None

    m = re.fullmatch(rf"({NUMBER})\s*(?:cm|c\.m|centimeters?|centimetres?)", t) or re.search(rf"\(({NUMBER})\s*cm\)", t)
    if m:
        cm = float(m.group(1))
    elif (m := re.fullmatch(rf"({NUMBER})\s*(?:inches|inche|inch|inc|in|\")", t)) and float(m.group(1)) >= 48:
        cm = float(m.group(1)) * 2.54
    elif re.fullmatch(NUMBER, t):
        v = float(t)
        if 100 <= v <= 250:                                       # bare number in cm
            cm = v
        elif re.fullmatch(r"\d\.(?:\d{3,}|1[2-9]|[2-9]\d)", t):   # 5.75 -> decimal feet (inches can't exceed 11)
            cm = v * 30.48
        elif not re.fullmatch(r"\d\.\d{1,2}", t):                 # bare '5' or '55' is not a height
            return None, "invalid"
    if cm is None and t.count(".") < 2:
        m = re.fullmatch(rf"(\d)\s*(?:{FT})?\s*[.,:\-]?\s*(?:{FT})?\s*(\d{{1,2}}(?:\.\d+)?)?\s*(?:{INCH}|{FT})?", t)
        if m:
            feet, inches = float(m.group(1)), float(m.group(2) or 0)
            if inches < 12:
                cm = (feet * 12 + inches) * 2.54
    if cm is None or not 120 <= cm <= 230:
        return None, "invalid"
    return round(cm, 1), kind


def clean_numeric(col, parser, unit, integer=False):
    """Strict numeric cleaning: parseable -> number, missing/unclear -> mode."""
    raw = df[col]
    ctype = detect_type(raw)
    parsed, kinds = [], []
    for v in raw:
        if is_missing_like(v):
            parsed.append(np.nan)
            kinds.append("missing")
        else:
            val, kind = parser(v)
            parsed.append(np.nan if val is None else val)
            kinds.append(kind)
    parsed = pd.Series(parsed, index=raw.index, dtype=float).round(2)
    kinds = pd.Series(kinds, index=raw.index)

    mode = parsed.mode().iloc[0]
    cleaned = parsed.fillna(mode)
    if integer:                                  # whole numbers only (e.g. age)
        cleaned, mode = cleaned.astype(int), int(mode)
    df[col] = cleaned
    log_cleaning(col, "numerical", raw, cleaned, (kinds == "missing").sum(), (kinds == "invalid").sum(),
                 kinds.isin(["converted", "range"]).sum(), mode)

    print(f"{col} | detected type: {ctype} | target unit: {unit}")
    print(f"missing replaced: {(kinds == 'missing').sum()} | invalid replaced: {(kinds == 'invalid').sum()} | "
          f"text/unit converted: {(kinds == 'converted').sum()} | ranges -> midpoint: {(kinds == 'range').sum()} | mode: {mode}")
    changes = pd.DataFrame({"before": raw, "after": cleaned, "action": kinds})
    display(changes[changes["action"].isin(["invalid", "converted", "range"])].sort_values("action"))
    display(cleaned.describe().round(2).to_frame().T)


# ---- one entry point for every column --------------------------------------
def make_whole_number_parser(parser):
    """Wrap a parser so the result is floored to whole units (20 year 8.5 month -> 20)."""
    def parse(value):
        amount, kind = parser(value)
        return (float(np.floor(amount)), kind) if amount is not None else (None, kind)
    return parse


# Column-specific settings; any column not listed here is cleaned as a plain categorical column.
NUMERIC_RULES = {   # column: (parser, unit, whole numbers only?)
    "A1_age": (make_whole_number_parser(make_quantity_parser(YEARS, min_value=10, max_value=100)), "years", True),
    "B2_family_time_hours": (make_quantity_parser(HOURS, min_value=0, max_value=24), "hours", False),
    "C1_height": (parse_height_cm, "cm", False),
    "C2_weight": (make_quantity_parser(KILOS, min_value=30, max_value=200), "kg", False),
    "D1_years_watching": (make_quantity_parser(YEARS, min_value=0, max_value=10), "years", False),
}
ALLOWED_VALUES = {"A2_gender": ["Male", "Female"]}


def clean_column(col):
    """Clean any column with the same rules: numeric columns use their parser, the rest are categorical."""
    if col in NUMERIC_RULES:
        parser, unit, integer = NUMERIC_RULES[col]
        clean_numeric(col, parser, unit, integer)
    else:
        clean_categorical(col, valid=ALLOWED_VALUES.get(col))
```
