```python
import pandas as pd
df = pd.read_csv("dataset/dataset_v1.csv")
df.head()
```

```python
df.dtypes
```

```python
# Group columns -> ordered categories (low to high)
group_order = {
    "A1_age_group": ["16-20", "21-25", "26-35"],
    "B2_family_time_group": ["Up to 1 hour", "1-3 hours", "3-5 hours", "5-8 hours", "More than 8 hours"],
    "C1_height_group": ["Up to 160 cm", "160-165 cm", "165-170 cm", "170-175 cm", "Above 175 cm"],
    "C2_weight_group": ["Up to 50 kg", "50-60 kg", "60-70 kg", "70-80 kg", "Above 80 kg"],
    "D1_years_group": ["Up to 1 year", "1-2 years", "2-3 years", "3-5 years", "More than 5 years"],
}
# Original column each group replaces (group takes its place)
group_source = {
    "A1_age_group": "A1_age",
    "B2_family_time_group": "B2_family_time_hours",
    "C1_height_group": "C1_height",
    "C2_weight_group": "C2_weight",
    "D1_years_group": "D1_years_watching",
}
group_cols = list(group_order)

for col, order in group_order.items():
    df[col] = pd.Categorical(df[col], categories=order, ordered=True)
    print(df[col].value_counts(sort=False))
    print()

# Put each group column where its original column was, and drop the original
new_cols = [c for c in df.columns if c not in group_cols]
new_cols = [next((g for g, s in group_source.items() if s == c), c) for c in new_cols]
df = df[new_cols]
df.columns.tolist()
```

```python
# Convert ALL columns to categorical (already-categorical columns keep their order)
df = df.astype("category")
df.dtypes.astype(str).value_counts()
```
